# Bag of Words (BoW) Representation and Term-Frequency Matrix

Here is the implementation of the Bag of Words (BoW) representation and the creation of a term-frequency matrix using Python's `scikit-learn` library.

## Task: 01: Bag of Words Matrix Construction (Customer Reviews Dataset)

**Scenario:** You are given three customer reviews. You need to construct a vocabulary and generate a term-frequency
matrix using CountVectorizer with English stop words removed.

**Instructions:**
1. Instantiate CountVectorizer(stop_words='english') and fit-transform the corpus.
2. Extract the vocabulary list using get_feature_names_out() .
3. Convert the transformed sparse matrix into a Pandas DataFrame for structured presentation.

````python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer

corpus = [
    "The product performance is amazing and fast",
    "The service was fast and performance was great",
    "Terrible customer service and bad performance"
]
````

````python
# Step 1: Instantiate and fit-transform
vectorizer = CountVectorizer(stop_words='english')
bow_matrix = vectorizer.fit_transform(corpus)
````

````python
# Step 2: Extract vocabulary
vocab = vectorizer.get_feature_names_out()
print("Vocabulary:", vocab)
````

````python
# Step 3: Convert to DataFrame
df = pd.DataFrame(bow_matrix.toarray(), columns=vocab)
print("\nTerm-Frequency Matrix:")
print(df)
````

## Task: 02: Document Search Engine & Relevance Ranking:

**Scenario:** Build a mini search engine that accepts a search query and ranks documents based on cosine similarity scores.

**Sample Input Documents & Query:**
```python


 documents = [
        "Machine learning algorithms analyze structured data effectively",
        "Deep learning and neural networks excel at processing unstructured data",
        "Natural language processing helps computers understand human language",
        "Python is widely used for machine learning and data science"
    ]
 query = ["machine learning algorithms for data"]


```

**Instructions:**
1. Fit `CountVectorizer` on `documents`.
2. Transform both `documents` and `query` into numerical vector arrays.
3. Compute pairwise cosine similarity using `cosine_similarity(query_vector, doc_vectors)`.
4. Display ranked documents from highest score to lowest score.

## Solution:

````python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity

documents = [
    "Machine learning algorithms analyze structured data effectively",
    "Deep learning and neural networks excel at processing unstructured data",
    "Natural language processing helps computers understand human language",
    "Python is widely used for machine learning and data science"
]

query = ["machine learning algorithms for data"]
````

````python
# Step 1: Fit CountVectorizer on documents
vectorizer = CountVectorizer()
doc_vectors = vectorizer.fit_transform(documents)
````

````python
# Step 2: Transform query using same vectorizer
query_vector = vectorizer.transform(query)

# Step 3: Compute cosine similarity between query and documents
similarity_scores = cosine_similarity(query_vector, doc_vectors)
````

````python
# Step 4: Rank documents based on similarity scores
scores = similarity_scores[0]
ranked_indices = scores.argsort()[::-1]

print("Query:", query[0])
print("\nRanked Documents:")
for rank, idx in enumerate(ranked_indices, start=1):
    print(f"Rank {rank}: (Score: {scores[idx]:.4f}) {documents[idx]}")
````
